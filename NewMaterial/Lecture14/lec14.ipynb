# Code part 1 - taking a potentiometer reading

Copy the following block of code into Thonny. It will cycle through a `while` loop, continually taking readings off of the `GP0` pin and printing them to your screen.

```python
from machine import ADC, Pin
import time

# Initialize ADC on pin 26 (for Raspberry Pi Pico)
potentiometer = ADC(Pin(26))
while True:
    # Read the ADC value (0-65535)
    pot_value = potentiometer.read_u16()
    print("Potentiometer value:", pot_value)
    time.sleep(0.1)
```

This brings in a new ingredient - the `ADC` function. ADC stands for analog-to-digital converter. The sensor itself outputs an analog waveform, but to work with that in our program we need to map that on to a number. ADC is a class that takes a Pin as input and has a `.read_u16()` function which converts the reading into an integer ranging from 0 (low) to 65535 (high).

# Code part 2 - Get data off of the Pico

Now we've got a sensor streaming through our program, but we need to get that off of the microcontroller and into a Colab notebook so we can do things like make plots and model the pendulum with an ODE. To do that you can run the following block of code. You'll need to add your single line of code to convert from a raw voltage value in `pot_value` into an angle measurement `theta_value`

```python
from machine import ADC, Pin
import time
# Initialize ADC on pin 26 (for Raspberry Pi Pico)
potentiometer = ADC(Pin(26))
# Open the file in write mode
with open("potentiometer_values.txt", "w") as file:
    file.write('dataList = [\n')
    while True:
        pot_value = potentiometer.read_u16() # Read the ADC value (0-65535)
        theta_value = 0 # Modify this line to convert from a raw voltage to an angle
        file.write(f"{theta_value},\n")
        file.flush() # Ensure the value is written to the file immediately
        print("Potentiometer/angle values:", pot_value,theta_value)
        time.sleep(0.1)
```

As soon as you start running this program, it will begin writing to the memory of the Pico. When you want to stop collecting data, hit the red stop sign to terminate the program. You can see the data stored in `potentiometer_values.txt` in Thonny.

To transfer transfer this file out of Thonny and onto your computer, select `View > Files > Right-click file > Download to computer`.
