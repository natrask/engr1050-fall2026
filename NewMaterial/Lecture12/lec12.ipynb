## Initializing numpy arrays for scalars, vectors and matrices

First we import the numpy library. It's a Python idiom to shorten numpy as `np`.

```python
import numpy as np
```

Essentially, you just wrap `np.array()` around a list. 

- **Scalar**: Use `np.array()` with a single value in a list.
    - Example: `a = np.array([5])`

- **Vector**: Use `np.array()` with a one-dimensional list.
    - Example: `v = np.array([1, 2, 3])`

- **Matrix**: Use `np.array()` with a list of lists, where each inner list represents a row.
    - Example: `m = np.array([[1, 2], [3, 4]])`

**Warning: whole numbers and decimals.** If you build an array out of whole numbers, numpy assumes it will only ever hold whole numbers. If you later store a decimal in it, numpy chops off everything after the decimal point, and it does not give you an error message.

There are two ways to avoid this:
- Write the numbers with a decimal point when you create the array: `1.0` instead of `1`.
- Convert an array you already have with `.astype(float)`.

Run the code below to see the problem and both fixes.

```python
# The problem: an array of whole numbers
v = np.array([1, 2, 3])
v[0] = 0.75
print(v)    # [0 2 3] - the 0.75 was chopped to 0

# Fix 1: write the numbers with a decimal point
v = np.array([1.0, 2.0, 3.0])
v[0] = 0.75
print(v)    # [0.75 2.   3.  ]

# Fix 2: convert an existing array with .astype(float)
v = np.array([1, 2, 3])
v = v.astype(float)
v[0] = 0.75
print(v)    # [0.75 2.   3.  ]
```

To extract elements, rows, or columns from a matrix:

- **Access an element:** Use `matrix[row, column]`.
    - Example: `m[1, 0]` gets the element in the second row, first column.

- **Access a row:** Use `matrix[row, :]`.
    - Example: `m[0, :]` gets the first row.

- **Access a column:** Use `matrix[:, column]`.
    - Example: `m[:, 1]` gets the second column.

- **Access a submatrix:** Use slicing `matrix[start_row:end_row, start_col:end_col]`.
    - Example: `m[0:2, 1:3]` gets rows 0-1 and columns 1-2.

- **Modify elements:** Assign values directly using indexing.
    - Example: `m[1, 1] = 10` sets the element in the second row, second column to 10.

- **Row operations:** Multiply a row by a number, add or subtract rows, and assign the result back to a row.
    - Example: `m[1, :] = m[1, :] - 2 * m[0, :]` replaces the second row with itself minus twice the first row.
