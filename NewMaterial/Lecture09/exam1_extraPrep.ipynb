   ```python
   range(stop)               # counts from 0 up to stop, not including stop

   range(start, stop, step)  # same, but jumps by step each time
   ```

   ```python
   list(range(5, 0, -1))     # [5, 4, 3, 2, 1]
   list(range(5, -1, -1))    # [5, 4, 3, 2, 1, 0]  (stop at -1 to include 0)
   ```

12. **A reminder about classes.** A class bundles together some data and the functions that work on that data. A simple class has three parts:

   - **The initializer**, `__init__`, which runs automatically every time you create a new object from the class. Its job is to set up the object's starting data.
   - **Variables** that start with `self.`, such as `self.color`. Each object gets its own copy, so two objects made from the same class can hold different values.
   - **Functions** defined inside the class. Their first input is always `self`, which lets them read and change that object's variables.

   ```python
   class Light:
       # The initializer runs automatically when you create a new Light
       def __init__(self, color):
           self.color = color      # variables that start with self. belong to this object
           self.isOn = False

       # A function inside the class; its first input is always self
       def switch_on(self):
           self.isOn = True

       def describe(self):
           return f"The {self.color} light is on: {self.isOn}"

   kitchenLight = Light("yellow")   # calls __init__ with color = "yellow"
   kitchenLight.switch_on()         # calls a function on this object
   print(kitchenLight.describe())   # The yellow light is on: True
   ```

14. **A reminder about list comprehensions.** A list comprehension is a one-line way to build a list with a `for` loop. The basic form is

   ```python
   newList = [expression for item in someList]
   ```

   which does the same thing as

   ```python
   newList = []
   for item in someList:
       newList.append(expression)
   ```

   There are three variations you should be able to read:

   ```python
   # Keep only some items: an if at the END filters which items go in
   [t for t in range(6) if t > 3]              # [4, 5]

   # Choose between two values: an if/else at the FRONT is part of the expression
   ['big' if t > 3 else 'small' for t in range(6)]
   # ['small', 'small', 'small', 'small', 'big', 'big']

   # Two for clauses act like a nested loop, with the first one on the outside
   [(a, b) for a in ['x', 'y'] for b in [1, 2]]
   # [('x', 1), ('x', 2), ('y', 1), ('y', 2)]
   ```

The following program is running on the Pico. Pressing button `K2` turns `LED2` on, and pressing it again turns `LED2` off.
```python
from machine import Pin
import time

class led:
    def __init__(self, pinnumber):
        self.pin = Pin(pinnumber, Pin.OUT)
        self.isOn = False
        self.off()

    def on(self):
        self.pin.value(1)
        self.isOn = True

    def off(self):
        self.pin.value(0)
        self.isOn = False

    def toggle(self):
        if self.isOn:
            self.off()
        else:
            self.on()

class button:
    def __init__(self, pinnumber, led):
        self.button = Pin(pinnumber, Pin.IN, Pin.PULL_DOWN)
        self.led = led
        # store the button value at start
        self.oldbuttonstate = self.button.value()

    def buttonPressed(self):
        self.led.toggle()

    def catchPress(self):
        newbuttonstate = self.button.value()
        if self.oldbuttonstate == 0 and newbuttonstate == 1:
            self.buttonPressed()
        self.oldbuttonstate = newbuttonstate
        time.sleep(0.01)

print("Program is running")
myLed = led(1)
myButton = button(27, myLed)
while True:
    myButton.catchPress()
```
