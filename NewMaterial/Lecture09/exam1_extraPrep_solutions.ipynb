````python
stringToCount = "It's a beautiful day for ENGR1050"

def count_vowels(s):
    count = 0
    for letter in s:
        # lower() makes 'E' count the same as 'e'
        if letter.lower() in 'aeiou':
            count = count + 1
    return count

print('Number of vowels:', count_vowels(stringToCount))
````

8. **A reminder about dictionaries.** A dictionary stores pairs of *keys* and *values*. You look things up by key instead of by position:

   ```python
   fruitCounts = {'apple': 3, 'banana': 5}   # key: value pairs inside curly braces

   print(fruitCounts['apple'])     # look up a value by its key: 3
   fruitCounts['apple'] += 1       # change a value: apple is now 4
   fruitCounts['cherry'] = 2       # add a new key

   if 'banana' in fruitCounts:     # check whether a key exists
       print('We have bananas')

   for fruit in fruitCounts:       # a for loop goes through the keys
       print(fruit, fruitCounts[fruit])
   ```

   Repeat the last exercise, but using a dictionary. You might use `vowelDict = {'a':0, 'e':0, 'i':0, 'o':0, 'u':0}` to get started.


````python
stringToCount = "It's a beautiful day for ENGR1050"

def count_each_vowel(s):
    vowelDict = {'a': 0, 'e': 0, 'i': 0, 'o': 0, 'u': 0}
    for letter in s:
        letter = letter.lower()
        if letter in vowelDict:          # only vowels are keys in the dictionary
            vowelDict[letter] += 1
    return vowelDict

vowelCounts = count_each_vowel(stringToCount)
print(vowelCounts)

vowelTotal = 0
for vowel in vowelCounts:
    vowelTotal = vowelTotal + vowelCounts[vowel]
print('Total number of vowels:', vowelTotal)
````

9. The factorial function is defined, for an integer n ≥ 0,

$$
n! = \prod_{k=1}^{n} k = 1\cdot 2\cdot 3\cdots n,
\qquad 0! = 1
$$

Define a function that implements this. Be sure that for the case `n=0` it returns 1 and otherwise evaluates the formula.


````python
def myfactorial(n):
    if n == 0:
        return 1
    output = 1
    for i in range(1, n + 1):      # multiply 1 * 2 * ... * n
        output = output * i
    return output

print('0! =', myfactorial(0))    # should be 1
print('5! =', myfactorial(5))    # should be 120
````

11. Write a function called `isPalindrome` that returns `True` if the input string is the same forward and backward.


````python
def isPalindrome(s):
    reversedString = ""
    for i in range(len(s) - 1, -1, -1):
        reversedString = reversedString + s[i]
    if s == reversedString:
        return True
    else:
        return False

print('Is "racecar" a palindrome?', isPalindrome("racecar"))   # True
print('Is "hello" a palindrome?', isPalindrome("hello"))       # False
````

12. **A reminder about classes.** A class bundles together some data and the functions that work on that data. A simple class has three parts:

   - **The initializer**, `__init__`, which runs automatically every time you create a new object from the class. Its job is to set up the object's starting data.
   - **Variables** that start with `self.`, such as `self.color`. Each object gets its own copy, so two objects made from the same class can hold different values.
   - **Functions** defined inside the class. Their first input is always `self`, which lets them read and change that object's variables.

   kitchenLight.switch_on()         # calls a function on this object
   print(kitchenLight.describe())   # The yellow light is on: True

````python
class BankAccount:
    def __init__(self, account_holder, balance=0):
        self.account_holder = account_holder
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print(f"Deposited ${amount}. New balance is ${self.balance}.")

    def withdraw(self, amount):
        # the only change: refuse the withdrawal if there is not enough money
        if amount > self.balance:
            print(f"Error: cannot withdraw ${amount}. Balance is only ${self.balance}.")
        else:
            self.balance -= amount
            print(f"Withdrew ${amount}. New balance is ${self.balance}.")

    def display_balance(self):
        print(f"Account holder: {self.account_holder}, Balance: ${self.balance}")

# Example code to use class
bobsAccount = BankAccount("Bob", 20)
bobsAccount.display_balance()
bobsAccount.deposit(50)
bobsAccount.withdraw(130)    # too much: prints the error, balance unchanged
bobsAccount.withdraw(30)     # fine
bobsAccount.display_balance()
````

14. **A reminder about list comprehensions.** A list comprehension is a one-line way to build a list with a `for` loop. The basic form is

   ```python
   newList = [expression for item in someList]
   ```

   which does the same thing as

   ```python
   newList = []
   for item in someList:
       newList.append(expression)
   ```

   There are three variations you should be able to read:

   ```python
   # Keep only some items: an if at the END filters which items go in
   [t for t in range(6) if t > 3]              # [4, 5]

   # Choose between two values: an if/else at the FRONT is part of the expression
   ['big' if t > 3 else 'small' for t in range(6)]
   # ['small', 'small', 'small', 'small', 'big', 'big']

   # Two for clauses act like a nested loop, with the first one on the outside
   [(a, b) for a in ['x', 'y'] for b in [1, 2]]
   # [('x', 1), ('x', 2), ('y', 1), ('y', 2)]
   ```

   What would be the output of each of the following list comprehensions?
```python
# 1
listcomp1 = [x*x for x in range(10)]          

# 2
nums = list(range(11))
listcomp2 = [n for n in nums if n % 2 == 0]     

# 3
words = ["apple", "banana", "cherry"]
listcomp3 = [w[0] for w in words]         

# 4 
listcomp4 = [(i, j) for i in range(3) for j in range(i+1, 3)]  

# 5
vals = [-2, -1, 0, 3]
listcomp5 = [v if v > 0 else 0 for v in vals]   
```


````python
# 1: the square of each number from 0 to 9
listcomp1 = [x*x for x in range(10)]
print(listcomp1)

# 2: keep only the even numbers from 0 to 10
nums = list(range(11))
listcomp2 = [n for n in nums if n % 2 == 0]
print(listcomp2)

# 3: the first letter of each word
words = ["apple", "banana", "cherry"]
listcomp3 = [w[0] for w in words]
print(listcomp3)

# 4: every pair (i, j) with i < j, from the numbers 0, 1, 2
listcomp4 = [(i, j) for i in range(3) for j in range(i+1, 3)]
print(listcomp4)

# 5: keep positive values, replace everything else with 0
vals = [-2, -1, 0, 3]
listcomp5 = [v if v > 0 else 0 for v in vals]
print(listcomp5)
````

**Explanation.** In #4 the two `for` clauses act as nested loops, with `i` on the outside. For each `i`, `j` runs from `i+1` up to 2, so `i = 0` gives two pairs, `i = 1` gives one, and `i = 2` gives none. In #5 the `if`/`else` sits at the front, so it chooses a value for every item instead of filtering items out. That is why the output has the same length as `vals`.

2. Write a class called `NumberList` that takes a list of numbers as the input to its initializer. It should have two functions:
    - `add_to_all(n)` - add the number `n` to every element of the list
    - `average()` - return the average of the numbers in the list

   You should be able to use it like this:
```python
myNumbers = NumberList([1, 2, 3, 4])
print(myNumbers.average())   # 2.5
myNumbers.add_to_all(10)
print(myNumbers.average())   # 12.5
```


````python
class NumberList:
    def __init__(self, numbers):
        self.numbers = numbers

    def add_to_all(self, n):
        # change each element in place, using its index
        for i in range(len(self.numbers)):
            self.numbers[i] = self.numbers[i] + n

    def average(self):
        total = 0
        for number in self.numbers:
            total = total + number
        return total / len(self.numbers)

myNumbers = NumberList([1, 2, 3, 4])
print(myNumbers.average())   # 2.5
myNumbers.add_to_all(10)
print(myNumbers.average())   # 12.5
````

````python
def mystery(x):
    if x >= 1:
        if x != 2:
            if x > 100:
                print("A")
            else:
                print("B")
        else:
            print("C")
    else:
        if x < 0:
            if x <= -1:
                print("D")
            else:
                print("E")
        else:
            if x == 0:
                print("F")
            else:
                print("G")

mystery(2.0)
mystery(-1)
mystery(1000)
````

- `mystery(1000)` prints **A**. `1000 >= 1`, `1000 != 2`, and `1000 > 100` are all `True`.

4. **A reminder about comparing strings.** The comparison operators `<`, `>`, `<=`, and `>=` work on characters, and they follow alphabetical order. For example:
```python
print('b' > 'a')   # True
print('a' >= 'c')  # False
```
They also work on whole strings, comparing one character at a time from the left, the way a dictionary is ordered. So `'Bryce Harper' < 'Trea Turner'` is `True` because `'B'` comes before `'T'`. One catch: every capital letter counts as coming before every lowercase letter, so `'Z' < 'a'` is `True`. All of the names below start with a capital, so you do not need to worry about that here.

   Write a function that takes a list of names as input and returns a new list with the names in alphabetical order. Use only lists, `for` loops, and `if` statements. Do not use any built-in sorting functions such as `sorted()` or `.sort()`.


````python
# The Phillies' starting lineup, in batting order
names = ['Kyle Schwarber', 'Trea Turner', 'Bryce Harper', 'Luis Arraez', 'Alec Bohm',
         'Bryson Stott', 'Brandon Marsh', 'J.T. Realmuto', 'Justin Crawford']

def alphabetize(nameList):
    # copy the names into a new list so the original is not changed
    result = []
    for name in nameList:
        result.append(name)

    # Bubble sort: compare neighbors and swap them if they are out of order.
    # After each pass, the largest remaining name has moved to the end.
    for i in range(len(result)):
        for j in range(len(result) - 1 - i):
            if result[j] > result[j + 1]:
                temp = result[j]
                result[j] = result[j + 1]
                result[j + 1] = temp
    return result

sortedNames = alphabetize(names)
for name in sortedNames:
    print(name)
````

   **Hint:** loop over the characters of each name one at a time, and use an `if` statement to check whether the character is a space, `' '`. Every character before the space belongs to the first name, and every character after it belongs to the last name. You can build up a string one character at a time with `+=`, just like adding to a number.


````python
# The Phillies' starting lineup, in batting order
names = ['Kyle Schwarber', 'Trea Turner', 'Bryce Harper', 'Luis Arraez', 'Alec Bohm',
         'Bryson Stott', 'Brandon Marsh', 'J.T. Realmuto', 'Justin Crawford']

firstNames = []
lastNames = []
for name in names:
    first = ""
    last = ""
    foundSpace = False
    for letter in name:
        if letter == ' ':
            foundSpace = True          # everything after this is the last name
        elif foundSpace:
            last = last + letter
        else:
            first = first + letter
    firstNames.append(first)
    lastNames.append(last)

print(firstNames)
print(lastNames)
````

8. This problem uses the `led` and `button` classes from the Lecture 11 lab. The board is wired the same way as in the last lab exercise, with all four LEDs and all four buttons connected:

| Jumper from | To | | Jumper from | To |
|---|---|---|---|---|
| `GP0` | `LED1` | | `GP28` | `K1` |
| `GP1` | `LED2` | | `GP27` | `K2` |
| `GP2` | `LED3` | | `GP26` | `K3` |
| `GP3` | `LED4` | | `GP22` | `K4` |

The following program is running on the Pico. Pressing button `K2` turns `LED2` on, and pressing it again turns `LED2` off.
```python
from machine import Pin
import time

class led:
    def __init__(self, pinnumber):
        self.pin = Pin(pinnumber, Pin.OUT)
        self.isOn = False
        self.off()

    def on(self):
        self.pin.value(1)
        self.isOn = True

    def off(self):
        self.pin.value(0)
        self.isOn = False

    def toggle(self):
        if self.isOn:
            self.off()
        else:
            self.on()

class button:
    def __init__(self, pinnumber, led):
        self.button = Pin(pinnumber, Pin.IN, Pin.PULL_DOWN)
        self.led = led
        # store the button value at start
        self.oldbuttonstate = self.button.value()

    def buttonPressed(self):
        self.led.toggle()

    def catchPress(self):
        newbuttonstate = self.button.value()
        if self.oldbuttonstate == 0 and newbuttonstate == 1:
            self.buttonPressed()
        self.oldbuttonstate = newbuttonstate
        time.sleep(0.01)

print("Program is running")
myLed = led(1)
myButton = button(27, myLed)
while True:
    myButton.catchPress()
```
   a. Which lines of the program decide that it is `K2` and `LED2`, and not some other button and LED? Explain how you can tell.

   b. Without changing any of the wiring, how would you change the program so that pressing `K1` toggles `LED4` instead? Only the code that creates the objects needs to change. You do not need to change either class.


**Answers.**

**a.** The two lines that create the objects:
```python
myLed = led(1)
myButton = button(27, myLed)
```
`led(1)` drives `GP1`, which the table shows is wired to `LED2`. `button(27, myLed)` reads `GP27`, which is wired to `K2`, and hands that button the LED it should control. Nothing inside the two classes mentions a specific pin. They get their pin numbers from these two lines.

**b.** Change only those two lines:
```python
myLed = led(3)                  # GP3 is wired to LED4
myButton = button(28, myLed)    # GP28 is wired to K1
```
Neither class changes. This is the payoff of passing the pin numbers to the constructor: changing which button controls which LED means changing two numbers.
