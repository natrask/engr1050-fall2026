## Example 9: Converting between types

```python
print(int("42"))
print(float("3.14"))
print(str(99))
print(bool(0), bool(7), bool(""), bool("no"))
```

## Example 10: Boolean operators

```python
hits = 150
walks = 80
on_base = hits + walks
print(on_base > 200 and hits > 100)
print(hits > 200 or walks > 50)
print(not (hits > 200))
```

## Example 11: Order of operations

## Example 12: Comparing floats

```python
print(0.1 + 0.2)
print(0.1 + 0.2 == 0.3)
print(abs((0.1 + 0.2) - 0.3) < 1e-9)
```

### Problem 1: predict the output of the four operators

```python
x = 5
y = 2
print(x / y, x // y, x % y, x ** y)
# 2.5, 2, 1, 25
# / is true division, // floor division, % modulo, ** exponent.
```

### Problem 2: Fahrenheit to Celsius converter

```python
def f_to_c(f):
    return (f - 32) * 5 / 9

# Demo with a hardcoded value (replace with input(...) when running yourself)
F = 100
print(f"{F} F is {f_to_c(F):.2f} C")
```

### Problem 3: the even-check bug

```python
# `if n % 2 = 0` is wrong because `=` is assignment.  Use `==`.
n = 8
if n % 2 == 0:
    print(True)
else:
    print(False)
```

### Problem 4: f-string with comma and space

```python
first = "Trea"
last = "Turner"
print(f"{last}, {first}")
```

### Problem 5: truthiness of falsy values

```python
print(bool(""), bool(" "), bool(0), bool(0.0), bool([]))
# False True False False False
# The empty string and the empty list are falsy, but " " (a space) is non-empty so it is truthy.
```
