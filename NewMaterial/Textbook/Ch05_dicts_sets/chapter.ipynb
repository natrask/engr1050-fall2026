## Example 1: dict as a lookup table

```python
rbi = {"Schwarber": 119, "Turner": 66, "Harper": 64, "Bohm": 64}
print(rbi["Harper"])
print(rbi["Bohm"])
```

## Example 2: get with a default

```python
rbi = {"Schwarber": 119, "Turner": 66, "Harper": 64, "Bohm": 64}
print(rbi.get("Realmuto"))
print(rbi.get("Realmuto", 0))
print(rbi.get("Harper", 0))
```

## Example 3: adding and updating

```python
rbi = {"Schwarber": 119, "Turner": 66, "Harper": 64, "Bohm": 64}
rbi["Stott"] = 55
rbi["Schwarber"] = 120
rbi.update({"Castellanos": 61, "Realmuto": 47})
print(rbi)
```

## Example 4: removing keys

```python
rbi = {"Schwarber": 119, "Turner": 66, "Harper": 64, "Bohm": 64, "Realmuto": 47}
removed = rbi.pop("Bohm")
print(removed)
print(rbi.get("Bohm"))
del rbi["Realmuto"]
print(rbi)
```

## Example 5: looping over items with formatting

```python
rbi = {"Schwarber": 119, "Turner": 66, "Harper": 64}
for name, n in rbi.items():
    print(f"{name:14s} {n:4d}")
```

## Example 6: histogram with a dict

```python
votes = ["pizza", "hoagies", "pizza", "pretzels", "hoagies", "hoagies", "pizza", "pizza"]
counts = {}
for v in votes:
    counts[v] = counts.get(v, 0) + 1
print(counts)
```

## Example 7: dict as a record

```python
player = {"name": "Trea Turner", "position": "SS", "rbi": 66, "hits": 165}
print(player["name"])
print(player["rbi"])
```

```python
roster = [
    {"name": "Schwarber", "rbi": 119},
    {"name": "Turner",    "rbi": 66},
    {"name": "Harper",    "rbi": 64},
]
for row in roster:
    print(row["name"], row["rbi"])
```

## Example 8: sorting a dict by value

```python
rbi = {"Schwarber": 119, "Turner": 66, "Harper": 64, "Bohm": 64}
top = sorted(rbi.items(), key=lambda kv: kv[1], reverse=True)
for name, n in top[:3]:
    print(name, n)
```

## Example 9: set basics

```python
states = {"PA", "NJ", "DE", "PA", "NJ"}
print(states)
print("PA" in states)
print(len(states))
```

## Example 10: dedupe two ways

```python
names = ["Trea", "Bryce", "Trea", "Kyle", "Bryce"]
print(list(set(names)))
print(list(dict.fromkeys(names)))
```

## Example 11: set operations

```python
team_a = {"Schwarber", "Turner", "Harper"}
team_b = {"Harper", "Bohm", "Realmuto"}
print(team_a | team_b)
print(team_a & team_b)
print(team_a - team_b)
```

## Example 12: dict comprehensions

```python
squares = {n: n * n for n in range(6)}
print(squares)

rbi = {"Schwarber": 119, "Turner": 66, "Harper": 64}
rbi_to_name = {v: k for k, v in rbi.items()}
print(rbi_to_name)
```

## Solutions

### Problem 1: count votes

```python
votes = ["yes", "no", "yes", "yes", "abstain", "no"]
counts = {}
for v in votes:
    counts[v] = counts.get(v, 0) + 1
print(counts)
```

### Problem 2: average via dict comprehension

```python
midterm = {"Alice": 88, "Bob": 75, "Cara": 92}
final   = {"Alice": 91, "Bob": 80, "Cara": 95}
total = {name: (midterm[name] + final[name]) / 2 for name in midterm}
print(total)
```

### Problem 3: lists do not support `&`, convert to sets first

```python
team_a = ["Schwarber", "Turner", "Harper"]
team_b = ["Harper", "Bohm", "Realmuto"]
shared = set(team_a) & set(team_b)
print(shared)
```

### Problem 4: character histogram

```python
s = "phillies"
counts = {}
for ch in s:
    counts[ch] = counts.get(ch, 0) + 1
print(counts)
```

### Problem 5: predict, then explain why list(d) is needed

```python
# Output: {'a': 1, 'c': 3}.  Even values were removed.
# list(d) takes a snapshot of the keys before the loop.
# Without it, mutating the dict during iteration would raise RuntimeError.
d = {"a": 1, "b": 2, "c": 3}
for k in list(d):
    if d[k] % 2 == 0:
        del d[k]
print(d)
```
