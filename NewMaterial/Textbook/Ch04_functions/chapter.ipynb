## Example 3: multiple parameters, positional and keyword

```python
def rectangle_area(width, height):
    return width * height

print(rectangle_area(4, 5))
print(rectangle_area(width=4, height=5))
print(rectangle_area(height=5, width=4))
```

## Example 4: default arguments

```python
def temperature_label(t, unit="F"):
    if unit == "F":
        return f"{t} degrees Fahrenheit"
    elif unit == "C":
        return f"{t} degrees Celsius"
    else:
        return f"{t} degrees ???"

print(temperature_label(72))
print(temperature_label(22, "C"))
print(temperature_label(295, unit="K"))
```

## Example 5: returning multiple values

```python
def descriptive_stats(values):
    n = len(values)
    mean = sum(values) / n
    spread = max(values) - min(values)
    return n, mean, spread

n, mean, spread = descriptive_stats([3, 5, 7, 9, 11])
print(n, mean, spread)
```

## Example 8: pure vs side-effecting

```python
def add_one(x):
    return x + 1

def add_one_to_each_pure(lst):
    return [x + 1 for x in lst]

print(add_one(7))
print(add_one_to_each_pure([1, 2, 3]))
```

## Example 9: the mutable-default trap

```python
def append_one_buggy(lst=[]):
    lst.append(1)
    return lst

print(append_one_buggy())
print(append_one_buggy())
print(append_one_buggy())
```

```python
def append_one(lst=None):
    if lst is None:
        lst = []
    lst.append(1)
    return lst

print(append_one())
print(append_one())
```

## Example 10: decomposing an ugly script

```python
def team_average(values):
    return sum(values) / len(values)

def top_performer(names, values):
    best_i = 0
    for i in range(1, len(values)):
        if values[i] > values[best_i]:
            best_i = i
    return names[best_i], values[best_i]

players = ["Schwarber", "Turner", "Harper"]
rbis = [119, 66, 64]
print(f"team RBI avg: {team_average(rbis):.1f}")
name, val = top_performer(players, rbis)
print(f"top RBI: {name} with {val}")
```

## Example 11: functions as values

```python
def apply(f, x):
    return f(x)

def double(x):
    return 2 * x

def negate(x):
    return -x

print(apply(double, 5))
print(apply(negate, 5))
```

## Solutions

### Problem 1: is_even and a comprehension

```python
def is_even(n):
    return n % 2 == 0

print([x for x in range(20) if is_even(x)])
```

### Problem 2: clamp

```python
def clamp(x, lo, hi):
    if x < lo:
        return lo
    if x > hi:
        return hi
    return x

print(clamp(150, 0, 100))
print(clamp(-3, 0, 100))
print(clamp(42, 0, 100))
```

### Problem 3: mutable-default bug, then a fixed version

```python
# The bug: history=[] is created once at def-time, shared across calls.
# So the third call sees [1, 2, 3] before it appends 3, total grows oddly.
# Better: take a history list from the caller.

def accumulate(value, history):
    history.append(value)
    return sum(history)

h = []
print(accumulate(1, h))
print(accumulate(2, h))
print(accumulate(3, h))
```

### Problem 4: stats returning three values

```python
def stats(values):
    return len(values), sum(values) / len(values), max(values)

n, m, big = stats([3, 5, 7])
print(n, m, big)
```

### Problem 5: apply_twice with lambda

```python
# f(y) = y + 3.  f(f(10)) = f(13) = 16.
def apply_twice(f, x):
    return f(f(x))
print(apply_twice(lambda y: y + 3, 10))
```
