## Example 1: counter

```python
class Counter:
    def __init__(self):
        self.value = 0
    def increment(self):
        self.value += 1

c = Counter()
c.increment(); c.increment(); c.increment()
print(c.value)
```

## Example 2: multiple instances are independent

```python
a = Counter(); b = Counter()
a.increment(); a.increment(); b.increment()
print(a.value, b.value)
```

## Example 3: rectangle with constructor

```python
class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height
    def area(self):
        return self.width * self.height
    def perimeter(self):
        return 2 * (self.width + self.height)

r = Rectangle(3, 5)
print(r.area())
print(r.perimeter())
```

## Example 4: bank account

```python
class Account:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance
    def deposit(self, amount):
        self.balance += amount
    def withdraw(self, amount):
        if amount > self.balance:
            print(f"insufficient funds for {self.owner}")
            return
        self.balance -= amount
    def show(self):
        print(f"{self.owner}: ${self.balance:.2f}")

acct = Account("Ben Franklin", 100)
acct.deposit(50)
acct.withdraw(30)
acct.show()
acct.withdraw(500)
```

## Example 5: refactoring into a stress-strain class

```python
class StressStrainData:
    def __init__(self, displacement, force):
        self.displacement = displacement
        self.force = force
    def zero_displacement(self):
        d0 = self.displacement[0]
        self.displacement = [d - d0 for d in self.displacement]
    def yield_estimate(self, fraction=0.95):
        peak = max(self.force)
        return peak * fraction

s = StressStrainData([2.0, 2.1, 2.3, 2.5], [0, 5, 10, 15])
s.zero_displacement()
print(s.displacement)
print(s.yield_estimate())
```

## Example 6: Circle

```python
import math
class Circle:
    def __init__(self, radius):
        self.radius = radius
    def area(self):
        return math.pi * self.radius * self.radius
    def circumference(self):
        return 2 * math.pi * self.radius

c = Circle(5)
print(c.area(), c.circumference())
```

## Example 7: defaults and clamping

```python
class Servo:
    def __init__(self, pin, neutral_angle=90):
        self.pin = pin
        self.angle = neutral_angle
    def move_to(self, angle):
        self.angle = max(0, min(180, angle))

s = Servo(pin=18)
s.move_to(250)
print(s.angle)
```

## Example 8: storing objects in objects

```python
class Roster:
    def __init__(self):
        self.players = []
    def add(self, player):
        self.players.append(player)
    def names(self):
        return [p.name for p in self.players]

class P:
    def __init__(self, name):
        self.name = name

r = Roster()
r.add(P("Schwarber")); r.add(P("Turner"))
print(r.names())
```

## Example 9: inheritance

```python
class Animal:
    def __init__(self, name):
        self.name = name
    def describe(self):
        return f"{self.name} is a {self.__class__.__name__}"
class Dog(Animal):
    def sound(self):
        return "woof"
class Cat(Animal):
    def sound(self):
        return "meow"

d = Dog("Fido"); c = Cat("Goldie")
print(d.describe(), d.sound())
print(c.describe(), c.sound())
```

## Example 10: override with super

```python
class Account:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance
    def deposit(self, amount):
        self.balance += amount

class FeeAccount(Account):
    def deposit(self, amount):
        super().deposit(amount * 0.99)

a = FeeAccount("Hamilton", 0)
a.deposit(100)
print(a.balance)
```

## Example 11: solver-style class (preview of Chapter 11)

```python
class ExplicitEuler:
    def __init__(self, f, t0, y0, tf, n):
        self.f = f
        self.t0 = t0
        self.y0 = y0
        self.tf = tf
        self.n = n
        self.h = (tf - t0) / n
    def solve(self):
        ts = [self.t0]
        ys = [self.y0]
        t = self.t0
        y = self.y0
        for _ in range(self.n):
            y = y + self.h * self.f(t, y)
            t = t + self.h
            ts.append(t)
            ys.append(y)
        return ts, ys

solver = ExplicitEuler(lambda t, y: -2 * y, 0, 1, 2, 20)
ts, ys = solver.solve()
print(ys[0], ys[-1])
```

## Example 12: __repr__

```python
class Account:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance
    def __repr__(self):
        return f"Account(owner={self.owner!r}, balance={self.balance})"

print(Account("Franklin", 50))
```

### Problem 1: Die class

```python
import random
class Die:
    def __init__(self, sides=6):
        self.sides = sides
    def roll(self):
        return random.randint(1, self.sides)

for d in [Die(), Die(20), Die(4)]:
    print(d.roll())
```

### Problem 2: Polynomial

```python
class Polynomial:
    def __init__(self, coeffs):
        self.coeffs = coeffs
    def eval(self, x):
        total = 0
        for i, c in enumerate(self.coeffs):
            total += c * x**i
        return total

print(Polynomial([1, 2, 3]).eval(2))
```

### Problem 4: SavingsAccount with interest

```python
class Account:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance
    def deposit(self, amount):
        self.balance += amount

class SavingsAccount(Account):
    def apply_interest(self, rate):
        self.balance += self.balance * rate

s = SavingsAccount("Trask", 1000)
s.apply_interest(0.05)
print(s.balance)
```

### Problem 5: Pendulum

```python
import math
class Pendulum:
    def __init__(self, length, mass):
        self.length = length
        self.mass = mass
    def period(self):
        return 2 * math.pi * math.sqrt(self.length / 9.81)

for L in [0.5, 1.0, 2.0]:
    print(L, Pendulum(L, 1).period())
```
