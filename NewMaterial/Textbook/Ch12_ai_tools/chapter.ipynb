## Example 2: reading generated code (the buggy yield-stress version)

````python
# This is the *buggy* version from the chapter, for reference.
# Run it on the example from the prompt and observe the wrong answer.
import numpy as np

def yield_stress_buggy(strain, stress):
    offset = 0.002
    diff = stress - 70000 * (strain - offset)
    idx = (diff < 0).argmax()
    return stress[idx]

strain = np.array([0.0, 0.001, 0.002, 0.003])
stress = np.array([0.0, 100.0, 150.0, 175.0])
print("buggy returns:", yield_stress_buggy(strain, stress))
print("expected:      ~137.5")
````

## Example 8: round trip on a normalize function

````python
import numpy as np

def normalize(x):
    span = x.max() - x.min()
    if span == 0:
        return np.zeros_like(x, dtype=float)
    return (x - x.min()) / span

print(normalize(np.array([2.0, 4.0, 8.0])))    # easy example
print(normalize(np.array([5.0, 5.0, 5.0])))    # degenerate, no nan
````

### Problem 2: is_prime edge cases

````python
# Below is a defensible version.  Verify each case.
def is_prime(n):
    if n < 2:
        return False
    if n < 4:
        return True
    if n % 2 == 0:
        return False
    k = 3
    while k * k <= n:
        if n % k == 0:
            return False
        k += 2
    return True

for n in [1, 2, 4, 9, 25, 0, -3]:
    print(n, is_prime(n))
````

### Problem 3: bug in the average function

````python
# The bug: `total + v` does not modify total; it computes a value and discards it.
# Use `total += v`.
def average(values):
    total = 0
    for v in values:
        total += v
    return total / len(values)

print(average([1, 2, 3, 4]))   # 2.5
````
