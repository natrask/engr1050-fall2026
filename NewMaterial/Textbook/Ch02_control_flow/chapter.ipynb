## Example 2: chained elif

```python
hour = 14
if hour < 12:
    print("morning")
elif hour < 17:
    print("afternoon")
elif hour < 21:
    print("evening")
else:
    print("night")
```

## Example 6: accumulating a sum

```python
total = 0
for n in range(1, 101):
    total += n
print(total)
```

```python
total = 0
for n in range(1, 101):
    if n % 2 == 0:
        total += n
print(total)
```

## Example 7: Collatz with while

```python
n = 100
steps = 0
while n > 1:
    if n % 2 == 0:
        n //= 2
    else:
        n = 3 * n + 1
    steps += 1
print(steps)
```

## Example 8: break with for-else

```python
players = ["Schwarber", "Turner", "Harper", "Bohm", "Castellanos"]
target = "Harper"
for name in players:
    if name == target:
        print(f"found {name}")
        break
else:
    print("not found")
```

## Example 9: continue

```python
for n in range(1, 11):
    if n % 3 == 0:
        continue
    print(n)
```

## Example 10: enumerate

```python
players = ["Schwarber", "Turner", "Harper"]
for i, name in enumerate(players):
    print(f"batter {i+1}: {name}")
```

## Example 11: zip

```python
players = ["Schwarber", "Turner", "Harper"]
rbis    = [119, 66, 64]
for name, rbi in zip(players, rbis):
    print(f"{name}: {rbi} RBI")
```

## Example 12: nested loops

```python
for i in range(1, 6):
    for j in range(1, 6):
        print(f"{i*j:4d}", end="")
    print()
```

## Example 13: validating a value with while


```python
candidates = ["abc", "-3", "0", "42"]
i = 0
while True:
    raw = candidates[i]
    if raw.isdigit() and int(raw) > 0:
        n = int(raw)
        break
    print(f"rejected {raw!r}")
    i += 1
print(f"accepted {n}")
```

### Problem 1: powers of 2

```python
for k in range(1, 11):
    print(f"2^{k} = {2**k}")
```

### Problem 2: count passing scores

```python
scores = [88, 92, 67, 100, 73, 81, 55]
passing = 0
for s in scores:
    if s >= 70:
        passing += 1
print(passing)
```

### Problem 3: only the first even > 100

```python
nums = [3, 17, 50, 75, 102, 200]
for n in nums:
    if n % 2 == 0 and n > 100:
        print(n)
        break
```

### Problem 4: predict the nested-loop output

```python
# Output: every (i, j) pair where i != j, for i,j in {1,2,3,4}.
# 12 pairs total.
for i in range(1, 5):
    for j in range(1, 5):
        if i == j:
            continue
        print(i, j)
```

### Problem 5: dot product with zip

```python
a = [1, 2, 3]
b = [4, 5, 6]
dot = 0
for x, y in zip(a, b):
    dot += x * y
print(dot)
```
