## Step 1: Make a class

I know that I want to have a `Gradebook` class whose job is to hold a list of students and their grades. To get started, I need to make a class.

The block of code below makes an empty class with an initializer that doesn't do anything yet (`pass` means "do nothing"). Remember that the `__init__` function is called automatically whenever a `Gradebook` is created. Our job will be to add variables and functions to this. Remember that there are examples in previous lectures' in-class exercises where you can copy and paste similar examples of a class to get started.

```python
class Gradebook:
    def __init__(self):
        pass
```

## Step 2: Add the internal variables that the class should hold

The gradebook should hold a list of students. Each student is stored as a dictionary, so the gradebook holds a list of dictionaries. For example, the first student in HW2 is listed as:

```python
firstStudentDictionary = {'name': 'John Adams', 'id': '2001', 'grades': [90, 92, 88, 91, 89]}
print("The first student's name is:", firstStudentDictionary['name'])
print("The first student's ID is:", firstStudentDictionary['id'])
print("The first student's grades are:", firstStudentDictionary['grades'])
```

We're going to add the list to the class by writing `self.studentList` inside `__init__`. Putting `self.` in front of a variable attaches it to one particular gradebook, so each `Gradebook` you create gets its own separate list of students.

```python
class Gradebook:
    def __init__(self):
        self.studentList = [] # this is the variable we added

myGradebook = Gradebook()
print(myGradebook.studentList) # so far we haven't added any students yet, so the list is empty
```

## Step 3: Add a function

OK - now let's add a function so that we can add students to the gradebook. We will copy and paste the class from above in the code block below, and add the first function from the homework.

The homework says `add_student(name, id)` should add a new student with the given name and ID number, and that the new student should start with an empty list of grades. So the function builds a new student dictionary out of the name and ID, and appends it to the student list.

Notice that the IDs in the homework data are strings like `'2001'`, with quotes, and not numbers. In Python `'2001'` and `2001` are not equal, so keep the quotes whenever you look a student up by ID.

```python
class Gradebook:
    def __init__(self):
        self.studentList = []

    def add_student(self, name, id):
        # a new student starts with no grades
        newStudent = {'name': name, 'id': id, 'grades': []}
        self.studentList.append(newStudent)

myGradebook = Gradebook()
myGradebook.add_student('John Adams', '2001')
print(myGradebook.studentList) # now the list contains the first student, with no grades yet
```

Note that the whole point of this is that we can make a second gradebook, with a totally separate list of students. For example:

```python
secondGradebook = Gradebook()
secondGradebook.add_student('Thomas Jefferson', '2024')
print('The first gradebook contains:')
print(myGradebook.studentList)
print('The second gradebook contains:')
print(secondGradebook.studentList)
```
