- In the code we wrote above, the command `led = Pin(25, Pin.OUT)` sets the pin to supply voltage as `25`. Pin number 25 is the internal LED pin. If you set it instead as `Pin(0, Pin.OUT)` it will apply voltage to `GP0`, `Pin(1, Pin.OUT)` to `GP1`, and so forth.

````python
from machine import Pin
import time

# Initialize which LED to blink
led = Pin(0, Pin.OUT) # THE ONLY CHANGE IS HERE: change 25 to 0

# Blink the LED continuously
while True:
    led.on()      # Turn LED on
    time.sleep(0.5)  # Wait 0.5 seconds
    led.off()     # Turn LED off
    time.sleep(0.5)  # Wait 0.5 seconds
````

# Making a button

In class we've collected user input in a few different ways: through command prompts and through GUIs. On the Pico, we are using a nice breadboard kit that provides some blue buttons as a rudementary way for a person to interact with the Pico. 

Collecting input on the Pico is a little bit tricky - whereas the `input` function we've used to collect input in Jupyter notebooks will wait for you to press enter to process the data, the Pico will run its `while` loop continuously without stopping, with each loop taking a fraction of a second unless you issue a `time.sleep` command. We will detect a button press by checking whether the button has flipped from on to off between steps.

````python
from machine import Pin
import time

# A class to catch when buttons are pressed and perform an action
class button:
    def __init__(self,pinnumber):

        self.button = Pin(pinnumber, Pin.IN, Pin.PULL_DOWN)

        # store the button value at start
        self.oldbuttonstate = self.button.value() 

    def buttonPressed(self):
        print("Button was pressed!")

    def catchPress(self):
    
        # check if any of the buttons have flipped their state        
        newbuttonstate = self.button.value()

        if self.oldbuttonstate == 0 and newbuttonstate == 1:
            # a button press was detected, now call the buttonPressed function
            self.buttonPressed()

        self.oldbuttonstate = newbuttonstate
        
        time.sleep(0.01)

# This is the main program that will run over and over
print("Program is running")
# make a button, using the constructor to tell it what pin it's attached to
myButton = button(28)
# now just keep checking the button state
while True:
    myButton.catchPress()
````

Take your time and read through this code. Talk to your group and make sure everyone understands what the deal is with `newbuttonstate` and `oldbuttonstate`. The logic of the code is the following
- At the bottom of the code, after we define the class, we have the program itself
- First we make an instance of a button, using the constructor to point it to the pin we have wired (mine was wired to `GP28` - which is why I called `button(28)`).
- Then over and over again, we call that button's `catchPress` function.
- When that function detects that the button as switched from off to on between two passes of the while loop, it will call the `self.buttonPressed()` function.
- In `buttonPressed()` we tell it what we want to have happen - that we print to the computer screen. You should see the following picture in your Thonny terminal, with a new "Button was pressed!" appearing everytime you press the button.
